c) Se considera el tramo AB como un tramo con dos movimientos, la un MRUA y luego, un MRU. El vehículo al tener una velocidad constante al llegar al punto B. La velocidad será obtenida por el momento al final del MRUA.

```python
# c) Vmrua = Vb

# v_b = vi + a * (tf-ti)
v_b = 0 + 3.5*(8-0)

print(f"La velocidad en el punto B es {v_b} m/s con dirección este")
```

```python
# d) Se considera el final de este tramo como un punto donde el vehículo se detiene desde tb=13s hasta tc=17s
from sympy import *
x = Symbol('x')
# 0 = v_b + a (tc - tb)
a = solveset(28 + x * (17 - 13), x)

print(f"La aceleración en el punto C es {a} m/s^2")
```

```python
solveset(28 + x * (17 - 13), x)
```

```python
# e) La posición en el punto B va a estar dada por las distancias de los movimientos MRUA + MRU.
# La variación de tiempo en el MRUA es 8s
# La variación de tiempo en el MRU es 5s

# X(t) = X_0 + V_0*t + 0.5*a*t**2
# v_b = 28 m/s
X_mrua = 0 + 0 + 0.5 * 3.5 * (8-0)**2

X_b = X_mrua + v_b * (13 - 8)

print(f" La posición en el punto B es {X_b} metros ")
```

```python
# f) La disntancia en el punto C esta dada por el MRUA de la fase desaceleración

# X(t) = X_0 + V_0*t + 0.5*a*t**2

X_c = X_b + 28 * (17-13) + 0.5 * (-7) * (17-13)**2

print(f" La posición en el punto C es {X_c} metros desde A")
```
